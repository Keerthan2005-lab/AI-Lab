```python
GOAL_STATE = (1, 2, 3, 8, 0, 4, 7, 6, 5)

def print_board(state):
    for i in range(3):
        print(state[i*3:i*3+3])
    print()

def find_empty(state):
    return state.index(0)
def get_possible_moves(state):
    empty_index = find_empty(state)
    row, col = empty_index // 3, empty_index % 3
    moves = []


    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in directions:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_empty_index = new_row * 3 + new_col
            new_state_list = list(state)

            new_state_list[empty_index], new_state_list[new_empty_index] = \
                new_state_list[new_empty_index], new_state_list[empty_index]

            moves.append(tuple(new_state_list))
    return moves

def dfs(current_state, path, visited):
    if current_state == GOAL_STATE:
        return path

    visited.add(current_state)

    for next_state in get_possible_moves(current_state):
        if next_state not in visited:
            result = dfs(next_state, path + [next_state], visited)
            if result:
                return result
    return None
tate
initial_state = (1, 2, 3, 0, 8, 4, 7, 6, 5)

print("Initial State:")
print_board(initial_state)


path = dfs(initial_state, [initial_state], set())

if path:
    print(f"Solution found in {len(path) - 1} steps:")
    for i, state in enumerate(path):
        print(f"Step {i}:")
        print_board(state)
else:
    print("No solution found for the given initial state within reasonable depth (due to recursion limits).")

print("Time Complexity : d^d")
```
